```python
import pandas as pd
index = pd.MultiIndex.from_tuples([
    ('Karnataka', 2010), ('Karnataka', 2020),
    ('Maharashtra', 2010), ('Maharashtra', 2020),
    ('Tamil Nadu', 2010), ('Tamil Nadu', 2020)
])
data = pd.Series([60, 67, 72, 112, 72, 77], index=index)
print(data)
```

```python
print(data.loc['Karnataka'])
print(data.xs(2020, level=1))
print(data.loc[('Karnataka', 2020)])
```

```python
df = data.unstack()
print(df)
s = df.stack()
print(s)
```

```python
import pandas as pd
import numpy as np
index = pd.MultiIndex.from_product([
    ['CSE', 'AIML'], [2023, 2024]
])
df = pd.DataFrame(np.random.randint(1, 100, (4, 2)),
                  index=index, columns=['A', 'B'])
print(df)
```

```python
import pandas as pd
import numpy as np
rows = pd.MultiIndex.from_product([
    ['Maths', 'Science'], ['Asha', 'Ravi']
], names=['subject', 'student'])
cols = pd.MultiIndex.from_product([
    ['Term1', 'Term2'], ['Internal', 'External']
], names=['term', 'exam_type'])
df = pd.DataFrame(np.random.randint(30, 100, (4, 4)),
                  index=rows, columns=cols)
print(df)
```

```python
print(df.loc['Maths'])
print(df.T.groupby(level='term').mean().T)
```

```python
import pandas as pd
index = pd.MultiIndex.from_tuples([
    ('StateB', 2020),
    ('StateA', 2010),
    ('StateC', 2020),
    ('StateB', 2010),
    ('StateA', 2020)
])
data = pd.Series([20, 10, 30, 15, 18], index=index)
try:
    print(data.loc['StateA':'StateB'])
except Exception as e:
    print("Error:", e)
data = data.sort_index()
print(data.loc['StateA':'StateB'])
```
