```python
import numpy as np
x = np.random.randint(1, 100, 10)
print("Original:", x)
y = np.sort(x)
print("np.sort():", y)
print("Original unchanged:", x)
x.sort()
print("x.sort():", x)
```

```python
import numpy as np
x = np.random.randint(1, 100, 10)
i = np.argsort(x)
print("Original:", x)
print("x[i]:", x[i])
print("np.sort(x):", np.sort(x))
```

```python
import numpy as np
x = np.random.randint(1, 100, (4, 6))
print("Original:\n", x)
print("Row sorted:\n", np.sort(x, axis=1))
print("Column sorted:\n", np.sort(x, axis=0))
```

```python
import numpy as np
x = np.random.randint(1, 100, 15)
p = np.partition(x, 4)
print("Original:", x)
print("4 smallest:", p[:4])
print("Verify:", np.sort(x)[:4])
```

```python
import numpy as np
student_names = np.array(["Rahul", "Anu", "Kiran", "Sneha", "Ravi"])
scores = np.array([75, 92, 68, 85, 78])
i = np.argsort(scores)
for j in i:
    print(student_names[j], scores[j])
```

```python
import numpy as np
X = np.random.rand(10, 2)
dist_sq = np.sum(
    (X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2,
    axis=-1
)
i = np.argsort(dist_sq, axis=1)
for j in range(10):
    print("Point", j, "->", i[j, 1:3])
```

```python
import numpy as np
dtype = [('title', 'U30'), ('author', 'U20'), ('price', 'f4')]
books = np.array([
    ('Python Basics', 'John', 450.0),
    ('Data Science', 'Smith', 600.0),
    ('Machine Learning', 'David', 750.0),
    ('NumPy Guide', 'James', 500.0)
], dtype=dtype)
print(books)
```

```python
print(books['title'])
print(books[1])
print(books[books['price'] < 600])
```

```python
dtype1 = {'names':['title','author','price'],
          'formats':['U30','U20','f4']}
dtype2 = [('title','U30'), ('author','U20'), ('price','f4')]
print(np.dtype(dtype1) == np.dtype(dtype2))
```

```python
data_rec = books.view(np.recarray)
print(data_rec.price)
```

```python
students = np.array([
    (1, 'Anu', 85),
    (2, 'Ravi', 92),
    (3, 'Kiran', 78),
    (4, 'Meena', 95),
    (5, 'Rahul', 88)
], dtype=[('roll_no','i4'), ('name','U20'), ('marks','f4')])
students = np.sort(students, order='marks')[::-1]
print(students)
```
