```python
import pandas as pd
df = pd.DataFrame({
    'department': ['CSE','AIML','ECE','CSE','AIML','ECE',
                   'CSE','AIML','ECE','CSE','AIML','ECE'],
    'gender': ['M','F','M','F','M','F','M','F','F','M','M','F'],
    'salary': [30000,40000,35000,45000,50000,38000,
               32000,42000,36000,33000,52000,39000]
})
print(pd.pivot_table(df, values='salary',
                     index='gender', columns='department',
                     aggfunc='mean'))
```

```python
print(pd.pivot_table(df, values='salary',
                     index='gender', columns='department',
                     aggfunc='count'))
```

```python
print(pd.pivot_table(df, values='salary',
                     index='gender', columns='department',
                     aggfunc='mean', margins=True))
```

```python
df['age'] = [22, 28, 35, 24, 42, 31,
             26, 29, 38, 23, 45, 33]
df['age_category'] = pd.cut(
    df['age'],
    bins=[0, 25, 35, 100],
    labels=['Junior', 'Mid', 'Senior']
)
print(pd.pivot_table(df, values='salary',
                     index='age_category',
                     columns='department',
                     aggfunc='mean', observed=False))
```

```python
print(pd.pivot_table(df, index='department',
                     values=['salary', 'age'],
                     aggfunc={'salary': 'sum', 'age': 'mean'}))
```
