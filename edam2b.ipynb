```python
import pandas as pd
import numpy as np
data = pd.Series([1, np.nan, 'hello', None])
print("Null values:")
print(data.isnull())
print("Non-null values:")
print(data[~data.isnull()])
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame(np.random.randint(1, 20, (5, 4)),columns=["A", "B", "C", "D"])
df.iloc[0, 1] = np.nan
df.iloc[1, 2] = np.nan
df.iloc[2, 0] = np.nan
df.iloc[3, 3] = np.nan
print(df)
print("\nMissing values:")
print(df.isnull().sum())
```

```python
print("After dropna():")
print(df.dropna())
print("\nAfter dropping columns:")
print(df.dropna(axis="columns"))
```

```python
df2 = pd.DataFrame({
    "A": [1, np.nan, 3, 4],
    "B": [2, np.nan, np.nan, 5],
    "C": [np.nan, np.nan, 6, 7]
})
print("Original:")
print(df2)
print("\nAfter dropna(how='all'):")
print(df2.dropna(how="all"))
```

```python
print("Rows having at least 3 non-null values:")
print(df2.dropna(thresh=3))
```

```python
import pandas as pd
import numpy as np
data = pd.Series([1, np.nan, 2, None, 3, np.nan])
print("Original:")
print(data)
print("\nFill with 0:")
print(data.fillna(0))
print("\nForward fill:")
print(data.ffill())
print("\nBackward fill:")
print(data.bfill())
```

```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Temperature": [30, 32, np.nan, 35, np.nan, 34, 33]
})
print("Original:")
print(df)
df_mean = df.copy()
df_mean["Temperature"] = df_mean["Temperature"].fillna(
    df_mean["Temperature"].mean()
)
print("\nUsing Mean:")
print(df_mean)
df_inter = df.copy()
df_inter["Temperature"] = df_inter["Temperature"].interpolate()
print("\nUsing Linear Interpolation:")
print(df_inter)
```
