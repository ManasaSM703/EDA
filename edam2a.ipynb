```python
import pandas as pd
population = pd.Series(
    [20, 19, 13, 12, 15],
    index=["Mumbai", "Delhi", "Bengaluru", "Chennai", "Kolkata"]
)
print("Bengaluru:", population["Bengaluru"])
print("\nDelhi to Chennai:")
print(population["Delhi":"Chennai"])
```

```python
import pandas as pd
area = pd.Series(
    [307713, 191791, 130060, 342239, 196024],
    index=["Karnataka", "Tamil Nadu", "Kerala", "Rajasthan", "Gujarat"]
)
population = pd.Series(
    [67, 77, 35, 81, 63],
    index=area.index
)
states = pd.DataFrame({
    "area": area,
    "population": population
})
states["density"] = states["population"] / states["area"]
print(states)
```

```python
import pandas as pd
indA = pd.Index([2, 3, 5, 7, 11])
indB = pd.Index([1, 3, 5, 7, 9])
print("Intersection:", indA.intersection(indB))
print("Union:", indA.union(indB))
print("Symmetric Difference:", indA.symmetric_difference(indB))
```
